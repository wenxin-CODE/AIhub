```python
# 字符串转数字，不使用api 
def stoi(s):
    num = 0
    for i in s:
        # 也可以使用ord代替下面
        # ord用于将单个字符转换为其对应的ASCII数值或Unicode数值。
        # 这个函数接受一个长度为1的字符串作为参数，并返回一个代表该字符的ASCII或Unicode编码的整数。
        num = num*10 + (ord(i) - ord('0'))
        # for j in range(10):
        #     if(i == str(j)):
        #         num = num*10+j

    return num

stoi("1234567890")
```
